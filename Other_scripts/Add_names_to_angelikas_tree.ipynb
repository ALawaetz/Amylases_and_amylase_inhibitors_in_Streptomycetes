```python
import pandas as pd
import re

# Load files
file1 = '/Users/andreaslawaetz/Streptomyces/Angeliga_project/SCOG/amylase_inhibitors/Master_table_with_amyPhyloGroups_known_clusters_and_inhibitors_GatorAcarviosin_GatorAcarbose_gatorValidamycinA_gatorBafilomycin_tendamisatPFAM_tendamisatSMART_tendamisatPFAM_gatorTrestatin_acarviosinMinReq.csv'
df = pd.read_csv(file1)

file2 = '/Users/andreaslawaetz/Streptomyces/Angeliga_project/SCOG/tip_labels_w_names.tsv'
names = pd.read_csv(file2, sep='\t')

# Function to convert strain name to tree format
def format_strain_name(name):
    # Remove only the long description
    name = name.replace('(high G+C Gram-positive bacteria)', '')

    # Remove parentheses but KEEP content inside
    name = re.sub(r'[()]', '', name)
    
    # Replace Streptomyces with S.
    name = name.replace("Streptomyces", "S.")
    
    # Replace spaces with underscores
    name = name.replace(" ", "_")
    
    # Remove double underscores
    name = re.sub(r"_+", "_", name)
    
    # Ensure trailing underscore
    if not name.endswith("_"):
        name = name + "_"
        
    return name

# Apply formatting
names["formatted_name"] = names["Name"].apply(format_strain_name)

# Create full tree tip label
names["new_tip_label"] = names["tip_label"] + "_" + names["formatted_name"]

# Merge into df
new_df = df.merge(names[["tip_label", "new_tip_label"]], on="tip_label", how="left")

# Replace old tip_label
new_df["tip_label"] = new_df["new_tip_label"]

# Drop helper column
new_df = new_df.drop(columns=["new_tip_label"])

# Save new metadata file
output_file = '/Users/andreaslawaetz/Streptomyces/Amylase_paper/Re_References_updated_2/Revisions/Master_table_with_amyPhyloGroups_known_clusters_and_inhibitors_GatorAcarviosin_GatorAcarbose_gatorValidamycinA_gatorBafilomycin_tendamisatPFAM_tendamisatSMART_tendamisatPFAM_gatorTrestatin_acarviosinMinReq_NAMES_UPDATED_APRIL_2026.csv'
new_df.to_csv(output_file, index=False)
```
